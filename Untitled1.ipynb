```python
!pip -q install pyarrow

import os, glob, zipfile, shutil, re
import pandas as pd
import numpy as np

ZIP_DIR = "/content/drive/MyDrive/ARTICULOS MIOS/ART. DEIS"
WORKDIR = "/content/work_sadu"
OUTDIR  = "/content/drive/MyDrive/ARTICULOS MIOS/ART. DEIS/PARQUET_SADU"

os.makedirs(WORKDIR, exist_ok=True)
os.makedirs(OUTDIR, exist_ok=True)

CAUSAS_INTERES = [2, 3, 4, 5, 10, 11, 30, 31]
COLS_KEEP = ["fecha", "IdCausa", "GLOSATIPOESTABLECIMIENTO", "Total"]

def extract_year(path):
    m = re.search(r"(19|20)\d{2}", os.path.basename(path))
    return int(m.group(0)) if m else None

zip_files = sorted(glob.glob(os.path.join(ZIP_DIR, "AtencionesUrgencia*.zip")))
years = [(extract_year(z), z) for z in zip_files]
years = [(y, z) for (y, z) in years if y is not None]
years = sorted(years, key=lambda x: x[0])

print("Años detectados:", [y for y, _ in years])

def process_year(year, zip_path):
    extract_path = os.path.join(WORKDIR, f"urg_{year}")
    os.makedirs(extract_path, exist_ok=True)

    with zipfile.ZipFile(zip_path, "r") as z:
        z.extractall(extract_path)

    csv_files = [f for f in os.listdir(extract_path) if f.lower().endswith(".csv")]
    if len(csv_files) != 1:
        raise RuntimeError(f"[{year}] Esperaba 1 CSV, encontré: {csv_files}")

    csv_path = os.path.join(extract_path, csv_files[0])

    parts = []
    reader = pd.read_csv(
        csv_path,
        sep=";",
        chunksize=500_000,
        encoding="latin1",
        low_memory=False
    )

    for chunk in reader:
        chunk.columns = [c.strip() for c in chunk.columns]
        missing = [c for c in COLS_KEEP if c not in chunk.columns]
        if missing:
            raise KeyError(f"[{year}] Faltan columnas: {missing}")

        chunk = chunk[chunk["IdCausa"].isin(CAUSAS_INTERES)][COLS_KEEP]
        chunk["Total"] = pd.to_numeric(chunk["Total"], errors="coerce").fillna(0).astype(np.int64)
        parts.append(chunk)

    df = pd.concat(parts, ignore_index=True)

    df["fecha"] = pd.to_datetime(df["fecha"], dayfirst=True, errors="coerce")
    df = df.dropna(subset=["fecha"])

    df_daily = (
        df.groupby(["fecha", "IdCausa", "GLOSATIPOESTABLECIMIENTO"], as_index=False)
          .agg(Total=("Total", "sum"))
    )

    out_path = os.path.join(OUTDIR, f"urgencias_respiratorias_{year}.parquet")
    df_daily.to_parquet(out_path, index=False, compression="snappy")

    shutil.rmtree(extract_path)
    return out_path, df_daily

log = []
parquet_paths = []

for y, zp in years:
    out_path, df_y = process_year(y, zp)
    parquet_paths.append(out_path)
    info = {
        "year": y,
        "zip": os.path.basename(zp),
        "rows": int(df_y.shape[0]),
        "min_date": str(df_y["fecha"].min().date()),
        "max_date": str(df_y["fecha"].max().date()),
        "total_sum": int(df_y["Total"].sum()),
        "parquet": out_path
    }
    log.append(info)
    print(f"[OK] {y} -> rows={info['rows']} total={info['total_sum']} parquet={os.path.basename(out_path)}")

# Unir todo en un solo parquet
df_all = pd.concat([pd.read_parquet(p) for p in parquet_paths], ignore_index=True)

final_path = os.path.join(OUTDIR, "urgencias_respiratorias_2017_2024.parquet")
df_all.to_parquet(final_path, index=False, compression="snappy")

log_df = pd.DataFrame(log)
log_csv = os.path.join(OUTDIR, "log_procesamiento_2017_2024.csv")
log_df.to_csv(log_csv, index=False)

print("\n✅ PARQUET FINAL:", final_path)
print("✅ LOG:", log_csv)
print("Filas:", df_all.shape[0])
print("Total atenciones (suma):", int(df_all["Total"].sum()))
print("Rango fechas:", df_all["fecha"].min(), "->", df_all["fecha"].max())
print("Tipos establecimiento:", df_all["GLOSATIPOESTABLECIMIENTO"].unique().tolist())
```

```python
def moving_zscore(series, window=14, threshold=3):
    mu = series.rolling(window, min_periods=window).mean()
    sd = series.rolling(window, min_periods=window).std()
    z = (series - mu) / sd
    alarm = z > threshold
    return mu, z, alarm

baseline_df["MA14"], baseline_df["Z"], baseline_df["Alarm_MA"] = zip(
    *baseline_df.groupby("Grupo")["Total"].apply(lambda s: moving_zscore(s, 14, 3))
)

baseline_df.head()
```

```python
import numpy as np
import pandas as pd

# Asegurar orden
baseline_df = baseline_df.sort_values(["Grupo","fecha"]).reset_index(drop=True)

# Media móvil y SD por grupo (ventana 14)
baseline_df["MA14"] = baseline_df.groupby("Grupo")["Total"].transform(
    lambda s: s.rolling(14, min_periods=14).mean()
)
baseline_df["SD14"] = baseline_df.groupby("Grupo")["Total"].transform(
    lambda s: s.rolling(14, min_periods=14).std()
)

# Z-score (evitar división por cero)
baseline_df["Z"] = (baseline_df["Total"] - baseline_df["MA14"]) / baseline_df["SD14"].replace(0, np.nan)

# Alarma
baseline_df["Alarm_MA"] = baseline_df["Z"] > 3

baseline_df[["fecha","Grupo","Total","MA14","Z","Alarm_MA"]].head(20)
```

```python
def cusum_pos(series, k=0.5, h=5):
    m = series.mean()
    s = np.zeros(len(series))
    alarm = np.zeros(len(series), dtype=bool)
    for i in range(1, len(series)):
        s[i] = max(0, s[i-1] + (series.iloc[i] - m - k))
        if s[i] > h:
            alarm[i] = True
            s[i] = 0
    return s, alarm

baseline_df["CUSUM"] = np.nan
baseline_df["Alarm_CUSUM"] = False

for g, idx in baseline_df.groupby("Grupo").groups.items():
    s = baseline_df.loc[idx, "Total"]
    s_pos, a = cusum_pos(s)
    baseline_df.loc[idx, "CUSUM"] = s_pos
    baseline_df.loc[idx, "Alarm_CUSUM"] = a

baseline_df[["fecha","Grupo","Total","CUSUM","Alarm_CUSUM"]].head(20)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12,5))

for g in ["APS","Hospital"]:
    tmp = baseline_df[baseline_df["Grupo"]==g].copy()
    ax.plot(tmp["fecha"], tmp["MA14"], label=f"{g} (MA14)")

    # marcar alarmas MA
    alarms = tmp[tmp["Alarm_MA"] & tmp["MA14"].notna()]
    ax.scatter(alarms["fecha"], alarms["MA14"], s=10, label=f"{g} alarmas Z>3")

ax.set_title("Baseline: media móvil 14 días + alarmas Z>3 (APS vs Hospital)")
ax.set_xlabel("Fecha")
ax.set_ylabel("Atenciones diarias (MA14)")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np

DATA_PATH = "/content/drive/MyDrive/ARTICULOS MIOS/ART. DEIS/PARQUET_SADU/urgencias_respiratorias_2017_2024.parquet"

df = pd.read_parquet(DATA_PATH)
df["fecha"] = pd.to_datetime(df["fecha"])

# Nacional total, causa respiratoria total (IdCausa=2)
y_df = (df[df["IdCausa"]==2]
        .groupby("fecha", as_index=False)
        .agg(y=("Total","sum"))
        .sort_values("fecha")
       )

y_df["dow"] = y_df["fecha"].dt.dayofweek  # 0=Lun ... 6=Dom
y = y_df["y"].astype(int).values
t = np.arange(len(y), dtype=float)  # índice temporal 0..n-1
n = len(y)

y_df.head(), y_df.tail(), n, y.sum()
```

```python
# Fourier anual (K=3 suele ir muy bien)
K = 3
period = 365.25

fourier = []
for k in range(1, K+1):
    fourier.append(np.sin(2*np.pi*k*t/period))
    fourier.append(np.cos(2*np.pi*k*t/period))
X_fourier = np.column_stack(fourier)  # n x (2K)

# One-hot día de semana (6 columnas, dejando una como referencia)
dow = y_df["dow"].values
X_dow = np.zeros((n, 6))
for j in range(6):
    X_dow[:, j] = (dow == j).astype(float)  # Dom (6) queda como referencia

X_fourier.shape, X_dow.shape
```

```python
import pymc as pm
import arviz as az
import pytensor.tensor as pt

with pm.Model() as model:
    # Priors
    intercept = pm.Normal("intercept", mu=np.log(np.mean(y)+1e-6), sigma=2.0)

    beta_fourier = pm.Normal("beta_fourier", mu=0.0, sigma=0.5, shape=X_fourier.shape[1])
    beta_dow = pm.Normal("beta_dow", mu=0.0, sigma=0.3, shape=X_dow.shape[1])

    # Change-point (tau) continuo (índice temporal)
    tau = pm.Uniform("tau", lower=0, upper=n-1)
    kappa = pm.HalfNormal("kappa", sigma=0.2)  # controla cuán abrupto es el cambio (más grande = más abrupto)
    delta = pm.Normal("delta", mu=0.0, sigma=1.0)  # cambio de nivel en log-escala

    # Switch suave: 0 antes de tau, 1 después (aprox)
    switch = pm.math.sigmoid(kappa * (t - tau))

    # Predicción media (log-link)
    eta = (intercept
           + pm.math.dot(X_fourier, beta_fourier)
           + pm.math.dot(X_dow, beta_dow)
           + delta * switch
          )

    mu = pm.Deterministic("mu", pm.math.exp(eta))

    # Sobredispersión NegBin
    alpha = pm.HalfNormal("alpha", sigma=5.0)

    # Likelihood
    y_obs = pm.NegativeBinomial("y_obs", mu=mu, alpha=alpha, observed=y)

    # Muestreo
    idata = pm.sample(
        draws=1500,
        tune=1500,
        chains=4,
        target_accept=0.9,
        random_seed=42
    )

az.summary(idata, var_names=["intercept","alpha","tau","kappa","delta"])
```

```python
import numpy as np
import pandas as pd
import pymc as pm
import arviz as az
import pytensor.tensor as pt

# --- Asegurar series por grupo ---
series = {}
for g in ["APS", "Hospital"]:
    sdf = (df[df["Grupo"]==g]
           .groupby("fecha", as_index=False)
           .agg(y=("Total","sum"))
           .sort_values("fecha"))
    series[g] = sdf

# --- Diseño estacional + DOW (mismo para ambos; largo igual para APS/Hospital porque cubren 2017-2024) ---
def build_design(sdf, K=3, period=365.25):
    y = sdf["y"].astype(int).values
    n = len(y)
    t = np.arange(n, dtype=float)

    # Fourier anual
    fourier = []
    for k in range(1, K+1):
        fourier.append(np.sin(2*np.pi*k*t/period))
        fourier.append(np.cos(2*np.pi*k*t/period))
    X_fourier = np.column_stack(fourier)  # n x (2K)

    # DOW one-hot (domingo referencia)
    dow = pd.to_datetime(sdf["fecha"]).dt.dayofweek.values
    X_dow = np.zeros((n, 6))
    for j in range(6):
        X_dow[:, j] = (dow == j).astype(float)

    return y, t, X_fourier, X_dow

# --- Fit por grupo (si no existe idata_by_group) ---
if "idata_by_group" not in globals():
    idata_by_group = {}

    for g in ["APS", "Hospital"]:
        sdf = series[g]
        y, t, X_fourier, X_dow = build_design(sdf, K=3)

        n = len(y)

        with pm.Model() as m:
            t_shared = pm.Data("t", t)

            intercept = pm.Normal("intercept", mu=np.log(np.mean(y)+1e-6), sigma=2.0)
            beta_fourier = pm.Normal("beta_fourier", 0.0, 0.5, shape=X_fourier.shape[1])
            beta_dow = pm.Normal("beta_dow", 0.0, 0.3, shape=X_dow.shape[1])

            # 2 change-points ordenados con min/max (compatibilidad)
            tau1_raw = pm.Uniform("tau1_raw", 0, n-1)
            tau2_raw = pm.Uniform("tau2_raw", 0, n-1)
            tau1 = pm.Deterministic("tau1", pt.minimum(tau1_raw, tau2_raw))
            tau2 = pm.Deterministic("tau2", pt.maximum(tau1_raw, tau2_raw))

            kappa = pm.HalfNormal("kappa", sigma=0.5)
            delta1 = pm.Normal("delta1", 0.0, 1.0)
            delta2 = pm.Normal("delta2", 0.0, 1.0)

            s1 = pm.math.sigmoid(kappa * (t_shared - tau1))
            s2 = pm.math.sigmoid(kappa * (t_shared - tau2))

            eta = (intercept
                   + pm.math.dot(X_fourier, beta_fourier)
                   + pm.math.dot(X_dow, beta_dow)
                   + delta1 * s1
                   + delta2 * s2)

            mu = pm.Deterministic("mu", pm.math.exp(eta))
            alpha = pm.HalfNormal("alpha", sigma=5.0)

            pm.NegativeBinomial("y_obs", mu=mu, alpha=alpha, observed=y)

            idata = pm.sample(
                draws=900, tune=900, chains=2,
                target_accept=0.92, random_seed=42
            )

        idata_by_group[g] = idata

print("Listo. Modelos guardados en idata_by_group:", list(idata_by_group.keys()))
```

```python
import numpy as np
import pandas as pd
import pymc as pm
import pytensor.tensor as pt
import arviz as az

# =========================
# CONFIG: fija tus quiebres nacionales aquí
# =========================
TAU1_DATE = pd.Timestamp("2020-03-28")  # <- τ1 nacional (COVID)
TAU2_DATE = pd.Timestamp("2021-10-31")  # <- τ2 nacional (post)
N_FOURIER = 3  # 3 armónicos -> 6 columnas (sin intercepto)
TARGET_ACCEPT = 0.90
DRAWS = 1500
TUNE = 1500
CHAINS = 4
SEED = 42

# =========================
# HELPERS
# =========================
def make_fourier_terms(dates, n_harm=3, period=365.25):
    """Crea matriz Fourier (sin intercepto): sin/cos para 1..n_harm"""
    t = (dates - dates.min()).dt.days.values.astype(float)
    X = []
    for k in range(1, n_harm + 1):
        X.append(np.sin(2*np.pi*k*t/period))
        X.append(np.cos(2*np.pi*k*t/period))
    return np.column_stack(X).astype(float), t

def make_dow_dummies(dates):
    """Dummies día de semana (6 columnas; referencia: domingo=6 o la última)"""
    dow = dates.dt.dayofweek.values  # 0=lun ... 6=dom
    X = np.zeros((len(dow), 6), dtype=float)
    # usamos 0..5 como dummies y 6 como referencia
    for i in range(6):
        X[:, i] = (dow == i).astype(float)
    return X

def date_to_index(dates, target_date):
    """Encuentra el índice más cercano en una serie diaria continua."""
    # asume dates es datetime64[ns]
    idx = int(np.argmin(np.abs((dates - target_date).values.astype("timedelta64[D]").astype(int))))
    return idx

def summarize_exp_delta(idata, name="delta"):
    v = idata.posterior[name].stack(sample=("chain","draw")).values
    expv = np.exp(v)
    med = float(np.median(expv))
    lo = float(np.percentile(expv, 5))
    hi = float(np.percentile(expv, 95))
    return med, lo, hi

# =========================
# MODELO: NegBin + estacionalidad + DOW + 2 change-points con taus FIJOS
# =========================
def fit_group_fixed_taus(df, tau1_date, tau2_date, group_name="APS"):
    sdf = df.copy()
    sdf["fecha"] = pd.to_datetime(sdf["fecha"])
    sdf = sdf.sort_values("fecha").reset_index(drop=True)

    y = sdf["y"].astype(int).values
    dates = sdf["fecha"]

    # matrices
    X_fourier, t = make_fourier_terms(dates, n_harm=N_FOURIER)
    X_dow = make_dow_dummies(dates)
    n = len(y)

    # índices fijos
    tau1_idx = date_to_index(dates, tau1_date)
    tau2_idx = date_to_index(dates, tau2_date)

    # tiempo como array pytensor-friendly
    t_arr = t.astype(float)

    with pm.Model() as model:
        # Datos
        Xf = pm.Data("Xf", X_fourier)
        Xd = pm.Data("Xd", X_dow)
        tt = pm.Data("tt", t_arr)

        # Priors
        intercept = pm.Normal("intercept", mu=np.log(np.mean(y) + 1e-6), sigma=2.0)
        beta_fourier = pm.Normal("beta_fourier", mu=0.0, sigma=0.5, shape=X_fourier.shape[1])
        beta_dow = pm.Normal("beta_dow", mu=0.0, sigma=0.3, shape=X_dow.shape[1])

        # Taus FIJOS (constantes)
        tau1 = pm.ConstantData("tau1_fixed", np.array(float(tau1_idx)))
        tau2 = pm.ConstantData("tau2_fixed", np.array(float(tau2_idx)))

        # Suavidad de transición (una por quiebre)
        kappa1 = pm.HalfNormal("kappa1", sigma=0.2)
        kappa2 = pm.HalfNormal("kappa2", sigma=0.2)

        # Cambios de nivel en log-escala
        delta1 = pm.Normal("delta1", mu=0.0, sigma=1.0)
        delta2 = pm.Normal("delta2", mu=0.0, sigma=1.0)

        # Switches suaves (0->1)
        sw1 = pm.math.sigmoid(kappa1 * (tt - tau1))
        sw2 = pm.math.sigmoid(kappa2 * (tt - tau2))

        eta = (intercept
               + pm.math.dot(Xf, beta_fourier)
               + pm.math.dot(Xd, beta_dow)
               + delta1 * sw1
               + delta2 * sw2)

        mu = pm.Deterministic("mu", pm.math.exp(eta))

        alpha = pm.HalfNormal("alpha", sigma=5.0)
        pm.NegativeBinomial("y_obs", mu=mu, alpha=alpha, observed=y)

        idata = pm.sample(
            draws=DRAWS, tune=TUNE, chains=CHAINS,
            target_accept=TARGET_ACCEPT, random_seed=SEED,
            progressbar=True
        )

    # Resumen para tabla
    d1_med, d1_lo, d1_hi = summarize_exp_delta(idata, "delta1")
    d2_med, d2_lo, d2_hi = summarize_exp_delta(idata, "delta2")

    out = {
        "Grupo": group_name,
        "tau1_fecha_fija": pd.Timestamp(tau1_date).date().isoformat(),
        "tau2_fecha_fija": pd.Timestamp(tau2_date).date().isoformat(),
        "tau1_idx": tau1_idx,
        "tau2_idx": tau2_idx,
        "exp_delta1_mediana": d1_med,
        "exp_delta1_ICr90_low": d1_lo,
        "exp_delta1_ICr90_high": d1_hi,
        "exp_delta2_mediana": d2_med,
        "exp_delta2_ICr90_low": d2_lo,
        "exp_delta2_ICr90_high": d2_hi,
    }
    return idata, pd.DataFrame([out])

# =========================
# EJECUCIÓN: APS y Hospital
# =========================
assert "series" in globals(), "No encuentro el dict `series` en memoria."
assert "APS" in series and "Hospital" in series, "series debe contener 'APS' y 'Hospital'."

idata_aps, tabla_aps = fit_group_fixed_taus(series["APS"], TAU1_DATE, TAU2_DATE, "APS")
idata_hosp, tabla_hosp = fit_group_fixed_taus(series["Hospital"], TAU1_DATE, TAU2_DATE, "Hospital")

TABLA_FIJA = pd.concat([tabla_aps, tabla_hosp], ignore_index=True)
TABLA_FIJA
```

```python
def fit_group_fixed_taus(df, tau1_date, tau2_date, group_name="APS"):
    sdf = df.copy()
    sdf["fecha"] = pd.to_datetime(sdf["fecha"])
    sdf = sdf.sort_values("fecha").reset_index(drop=True)

    y = sdf["y"].astype(int).values
    dates = sdf["fecha"]

    # matrices
    X_fourier, t = make_fourier_terms(dates, n_harm=N_FOURIER)
    X_dow = make_dow_dummies(dates)
    n = len(y)

    # índices fijos
    tau1_idx = date_to_index(dates, tau1_date)
    tau2_idx = date_to_index(dates, tau2_date)

    # convertir a float (constantes)
    tau1 = float(tau1_idx)
    tau2 = float(tau2_idx)

    with pm.Model() as model:
        # Datos
        Xf = pm.Data("Xf", X_fourier)
        Xd = pm.Data("Xd", X_dow)
        tt = pm.Data("tt", t.astype(float))

        # Priors
        intercept = pm.Normal("intercept", mu=np.log(np.mean(y) + 1e-6), sigma=2.0)
        beta_fourier = pm.Normal("beta_fourier", mu=0.0, sigma=0.5, shape=X_fourier.shape[1])
        beta_dow = pm.Normal("beta_dow", mu=0.0, sigma=0.3, shape=X_dow.shape[1])

        # Suavidad
        kappa1 = pm.HalfNormal("kappa1", sigma=0.2)
        kappa2 = pm.HalfNormal("kappa2", sigma=0.2)

        # Cambios de nivel
        delta1 = pm.Normal("delta1", mu=0.0, sigma=1.0)
        delta2 = pm.Normal("delta2", mu=0.0, sigma=1.0)

        # Switches con τ FIJOS
        sw1 = pm.math.sigmoid(kappa1 * (tt - tau1))
        sw2 = pm.math.sigmoid(kappa2 * (tt - tau2))

        eta = (intercept
               + pm.math.dot(Xf, beta_fourier)
               + pm.math.dot(Xd, beta_dow)
               + delta1 * sw1
               + delta2 * sw2)

        mu = pm.Deterministic("mu", pm.math.exp(eta))

        alpha = pm.HalfNormal("alpha", sigma=5.0)
        pm.NegativeBinomial("y_obs", mu=mu, alpha=alpha, observed=y)

        idata = pm.sample(
            draws=DRAWS,
            tune=TUNE,
            chains=CHAINS,
            target_accept=TARGET_ACCEPT,
            random_seed=SEED
        )

    # resumen para tabla
    d1_med, d1_lo, d1_hi = summarize_exp_delta(idata, "delta1")
    d2_med, d2_lo, d2_hi = summarize_exp_delta(idata, "delta2")

    out = {
        "Grupo": group_name,
        "tau1_fecha_fija": tau1_date.date().isoformat(),
        "tau2_fecha_fija": tau2_date.date().isoformat(),
        "tau1_idx": tau1_idx,
        "tau2_idx": tau2_idx,
        "exp_delta1_mediana": d1_med,
        "exp_delta1_ICr90_low": d1_lo,
        "exp_delta1_ICr90_high": d1_hi,
        "exp_delta2_mediana": d2_med,
        "exp_delta2_ICr90_low": d2_lo,
        "exp_delta2_ICr90_high": d2_hi,
    }

    return idata, pd.DataFrame([out])
```
